```python
# Crea una función que reciba una secuencia y devuelva el porcentaje GC.
def estimation(seq):
    seq = input("escribe tu secuencia:")
    c = (seq.upper()).count("C")
    print(f"el numero de Cs es {c}")
    g = (seq.upper()).count("G")
    print(f"el numero de Gs es {g}")
    t = len(seq.upper())
    print(f"el numero de bases es {t}")
    cg = ((c + g)/t) * 100
    return print(f"el % CG es '{cg}'")

estimation(seq)
```

```python
# Escriba un código de Python para determinar si la secuencia de ADN
# ADNseq se cortará con cualquiera de estas enzimas de restricción.
# Imprima la ubicación (índice) de los sitios de reconocimiento en la
# secuencia de ADN. Para este problema, debe recorrer las claves de la
# variable diccionario.

# Secuencia de ADN proporcionada
ADNseq = "GGCGATGCTAGTCGCGTAGTCTTAAGCTTGTCGAGAATTCGGATGTCATGA"

# Diccionario de enzimas y sus sitios de reconocimiento
restriction_enzymes = {"EcoRI" : "GAATTC", 
                        "AluI" : "AGCT", 
                        "NotI" : "GCGGCCGC", 
                        "TaqI" : "TCGA"}

for site in restriction_enzymes:
    a1 = restriction_enzymes.get(site)
    print(f"testing enzyme {site} with motif {a1}")
    position = ADNseq.find(a1)
    if position != -1:
        print(f"enzyme {site} cuts in position {position}")
    else:
        print(f"enzyme {site} does NOT cuts sequence")
    print("\n")
```
